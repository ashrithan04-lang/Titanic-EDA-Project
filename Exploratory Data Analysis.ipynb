```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("titanic-2024-02-08.xlsx")

df.head()
```

```python
sns.countplot(data=df, x="Pclass", hue="Survived")

plt.title("Titanic Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived", labels=["No", "Yes"])

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Age",
    hue="Survived",
    bins=20,
    multiple="stack"
)

plt.title("Passenger Age Distribution by Survival")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
survival_by_gender = df.groupby("Sex")["Survived"].mean() * 100

print(survival_by_gender)
```

```python
survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100

print(survival_by_class)
```

```python
correlation = df[["Age", "Fare", "Survived", "Pclass"]].corr()

print(correlation)
```

```python
survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100

survival_by_class.plot(kind="bar")

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
survival_count = df["Survived"].value_counts()

print("Passenger survival count:")
print(survival_count)

print("\nSurvival percentage:")
print((df["Survived"].value_counts(normalize=True) * 100).round(2))
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
print("Dataset shape:", df.shape)
```

```python

df["Age"] = df["Age"].fillna(df["Age"].median())


df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])


df = df.drop(columns=["Cabin"])


print(df.isnull().sum())
```

```python
print("Dataset shape after cleaning:", df.shape)
print("\nMissing values:")
print(df.isnull().sum().sum())
print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
survival_by_port = df.groupby("Embarked")["Survived"].mean() * 100

print("Survival rate by boarding location:")
print(survival_by_port.round(2))
```

```python
survival_by_port.plot(kind="bar")

plt.title("Survival Rate by Boarding Location")
plt.xlabel("Boarding Location")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python

df["FamilySize"] = df["SibSp"] + df["Parch"] + 1


survival_by_family = df.groupby("FamilySize")["Survived"].mean() * 100

print("Survival rate by family size:")
print(survival_by_family.round(2))
```

```python
survival_by_family.plot(kind="bar")

plt.title("Survival Rate by Family Size")
plt.xlabel("Family Size")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
average_fare = df.groupby("Pclass")["Fare"].mean()

print("Average ticket fare by passenger class:")
print(average_fare.round(2))
```

```python
average_fare.plot(kind="bar")

plt.title("Average Ticket Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.xticks(rotation=0)
plt.show()
```
